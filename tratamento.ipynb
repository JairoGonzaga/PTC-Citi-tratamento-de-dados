```python
import pandas as pd
```

```python
csv_path = "base_desp.csv"
df = pd.read_csv(csv_path)
df.head()
```

```python
#tratamento da coluna de sexo
df['sexo'] = df['sexo'].str.replace(r'^[Mm].*', 'M', regex=True)
df['sexo'] = df['sexo'].str.replace(r'^[Ff].*', 'F', regex=True)
```

```python
df.head() #verificação
```

```python
# para contas, em py o ',' nao pode ser usado, entao mudei para '.' para depois retornar para ','
df['nota_matematica']=df['nota_matematica'].str.replace(',','.',regex=False)
df['nota_portugues']=df['nota_portugues'].str.replace(',','.',regex=False)
df.head()
```

```python
df['nota_matematica'] = pd.to_numeric(df['nota_matematica'], errors='coerce')
df['nota_portugues'] = pd.to_numeric(df['nota_portugues'], errors='coerce')
df['frequencia'] = pd.to_numeric(df['frequencia'], errors='coerce')
df['media_notas'] = ((df['nota_matematica'] + df['nota_portugues'] + (df['frequencia'] / 10)) / 3 ).round(1)# criação da coluna media
```

```python
df.head()
```

```python
df['aprovado'] = df['media_notas'].map(lambda x: 'Aprovado' if x > 7.0 else 'Reprovado') #criação da coluna aprovados
```

```python
df.head()
```

```python
#transformo tudo para string
df['nota_matematica'] = df['nota_matematica'].astype(str)
df['nota_portugues'] = df['nota_portugues'].astype(str)
df['frequencia'] = df['frequencia'].astype(str)
df['media_notas'] = df['media_notas'].astype(str)

#formatação dos numeros de '.' para ','
df['nota_matematica']=df['nota_matematica'].str.replace('.',',',regex=False)
df['nota_portugues']=df['nota_portugues'].str.replace('.',',',regex=False)
df['frequencia']=df['frequencia'].str.replace('.',',',regex=False)
df['media_notas']=df['media_notas'].str.replace('.',',',regex=False)
df.head()
```

```python
df.to_csv('base_pronta.csv', index=False) #salvando alterações
```
